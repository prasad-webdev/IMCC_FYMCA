```python
# 1. Read a file containing student names (one per line) and store them in a list.
file = open("students1.txt", "r")
names = file.read().split()
file.close()
print(names)
```

```python
#2. Print names longer than 5 characters. 
for name in names:
    if len(name) > 5:
        print(name)
```

```python
#3. Count how many names start with a vowel.
count = 0
for name in names:
    if name[0].lower() in "aeiou":
        count += 1
print(count)
```

```python
#4. Save the filtered names (longer than 5 characters) into a new file. 
names1 = ["Rohan", "Pooja", "Vikramaditya", "Ananya", "Raj"]
with open("filtered_names.txt", "w") as file:
    for name in names:
        clean_name = name.strip()
        if len(clean_name) > 5:
            file.write(clean_name)
```

```python
#5. Sort the list of names alphabetically and print them in ascending order. 
sorted_list = sorted(names)
print(sorted_list)
```

```python
#6. Print names that contain the letter "a". 
for name in names:
    if "a" in name.lower():
        print(name)
```

```python
#7. Count word frequency in names (handle duplicates) and store results in a dictionary {name: count}. 
from collections import Counter
frequency = Counter(name.lower() for name in names)
print(dict(frequency))
```
