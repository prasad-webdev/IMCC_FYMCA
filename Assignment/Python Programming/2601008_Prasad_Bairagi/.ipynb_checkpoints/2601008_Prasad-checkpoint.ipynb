```python
#6.Seek Practice Jump to the 10th character using seek() and print the next 5 characters.
file = open("myfile.txt","r")
file.seek(10)
print(file.read(5))
file.close()
```

```python
#7.Line Numbering Print each line with its line number (like 1: First line).
file = open("myfile.txt","r")
num = 1
for line in file:
    print(num,":",line)
    num = num + 1
file.close()
```

```python
#8.Reverse File Content Read all lines and write them into another file in reverse order.
file = open("myfile.txt", "r")
lines = file.read().splitlines()
file.close()
file = open("reverse.txt", "w")
for line in lines[::-1]:
    file.write(line[::-1] + "\n")
file.close()
```

```python
#9.Word Frequency Counter Count how many times each word appears in a file and display results as a dictionary.
file = open("myfile.txt","r")
words = file.read().split()
file.close()
frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1
print(frequency)
```

```python
#10.File Comparison Compare two files and print whether their contents are identical
f1 = open("myfile.txt", "r")
f2 = open("sample.txt", "r")
if f1.read() == f2.read():
    print("Files are identical")
else:
    print("Files are different")
f1.close()
f2.close()
```

```python
#11.Copy Selected Lines Copy only even-numbered lines from one file into another.
f1 = open("myfile.txt", "r")
lines = f1.readlines()
f1.close()
f2 = open("even.txt", "w")
for i, line in enumerate(lines, 1):
    if i % 2 == 0:
        f2.write(line)
f2.close()
```
